```python
import numpy as np

cities = np.array([
    "Tehran", "Tabriz", "Shiraz", "Mashhad", "Kashan",
    "Rasht", "Esfehan", "Kerman", "Yazd", "Ramsar"
])

population = np.array([
    8693706, 1558693, 1565572, 3001184, 304000,
    679995, 1961260, 537718, 529673, 35000
])

latitude = np.array([
    35.69439, 38.08000, 29.61031, 36.29807, 33.98500,
    37.27611, 32.65246, 30.28321, 31.89722, 36.90336
])

longitude = np.array([
    51.42151, 46.29190, 52.53113, 59.60567, 51.41000,
    49.58862, 51.67462, 57.07879, 54.36750, 50.65858
])
```

```python
plt.scatter(longitude, latitude)

for i in range(len(cities)):
    plt.text(longitude[i], latitude[i], cities[i])

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Cities Location")

plt.show()
```

```python
plt.bar(cities, population, color='green')
plt.xticks(rotation=60)
plt.xlabel("City")
plt.ylabel("People")
plt.title("Population of Cities")
plt.show()
```

```python
plt.plot(cities, population,color= 'red')
plt.xticks(rotation=60)
plt.xlabel("City")
plt.ylabel("People")
plt.title("Population Chart")
plt.show()
```

```python
plt.pie(population, labels=cities, autopct='%1.2f%%')
plt.title("Population of Cities")
plt.show()
```
